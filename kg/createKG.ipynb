```python
g = Graph()
n = Namespace('https://koncordantlab.com/TTEXTS/')
genre = Namespace('https://koncordantlab.com/TTEXTS/genre/')
theme = Namespace('https://koncordantlab.com/TTEXTS/theme/')
subtheme = Namespace('https://koncordantlab.com/TTEXTS/subtheme/')

g.bind('TTEXTS',n)
g.bind('genre', genre)
g.bind('theme', theme)
g.bind('subtheme', subtheme)
```

```python
Text = URIRef(n.Text)
Author = URIRef(n.Author)

Text_Complexity = URIRef(n.Text_Complexity)

Qualitative_Measures = URIRef(n.Qualitative_Measures)
Levels_of_Meaning = URIRef(n.Levels_of_Meaning)
Text_Structure = URIRef(n.Text_Structure)
Language_Conventionality_and_Clarity = URIRef(n.Language_Conventionality_and_Clarity)
Knowledge_Demands = URIRef(n.Knowledge_Demands)

Exceedingly_Complex = URIRef(n.Exceedingly_Complex)
Very_Complex = URIRef(n.Very_Complex)
Moderately_Complex = URIRef(n.Moderately_Complex)
Slightly_Complex = URIRef(n.Slightly_Complex)

has_levels_of_meaning = URIRef(n['has_levels_of_meaning'])
has_text_structure = URIRef(n['has_text_structure'])
has_language_conventionality_and_clarity = URIRef(n['has_language_conventionality_and_clarity'])
has_knowledge_demands = URIRef(n['has_knowledge_demands'])

Quantitative_Measures = URIRef(n.Quantitative_Measures)
Lexile_Level = URIRef(n.Lexile_Level)
Flesch_Kincaid = URIRef(n.Flesch_Kincaid)
ATOS_Reading_Level = URIRef(n.ATOS_Reading_Level)
Reader_Maturity = URIRef(n.Reader_Maturity)
Complexity_Measurement = URIRef(n.Complexity_Measurement)

has_qualitative_measures = URIRef(n['has_qualitative_measures'])
has_quantitative_measures = URIRef(n['has_quantitative_measures'])
has_text_complexity = URIRef(n['has_text_complexity'])
has_lexile_level = URIRef(n['has_lexile_level'])
has_flesch_kincaid = URIRef(n['has_flesch_kincaid'])
has_atos_reading_level = URIRef(n['has_atos_reading_level'])
has_reader_maturity = URIRef(n['has_reader_maturity'])
has_min_reading_maturity = URIRef(n['has_min_reading_maturity'])
has_max_reading_maturity = URIRef(n['has_max_reading_maturity'])
has_complexity_measurement = URIRef(n['has_complexity_measurement'])

g.add((has_lexile_level, RDFS.range, XSD.integer))
g.add((has_flesch_kincaid, RDFS.range, XSD.decimal))
g.add((has_atos_reading_level, RDFS.range, XSD.decimal))
g.add((has_max_reading_maturity, RDFS.range, XSD.integer))
g.add((has_min_reading_maturity, RDFS.range, XSD.integer))

g.add((has_lexile_level, RDF.type, OWL.DatatypeProperty))
g.add((has_flesch_kincaid, RDF.type, OWL.DatatypeProperty))
g.add((has_atos_reading_level, RDF.type, OWL.DatatypeProperty))
g.add((has_reader_maturity, RDF.type, OWL.ObjectProperty))
g.add((has_max_reading_maturity, RDF.type, OWL.DatatypeProperty))
g.add((has_min_reading_maturity, RDF.type, OWL.DatatypeProperty))

has_pages = URIRef(n['has_pages'])
has_summary = URIRef(n['has_summary'])
has_year_of_publication = URIRef(n['has_year_of_publication'])
g.add((has_pages, RDF.type, OWL.DatatypeProperty))
g.add((has_pages, RDFS.range, XSD.integer))
g.add((has_summary, RDF.type, OWL.DatatypeProperty))
g.add((has_summary, RDFS.range, XSD.string))
g.add((has_year_of_publication, RDF.type, OWL.DatatypeProperty))
g.add((has_year_of_publication, RDFS.range, XSD.gYear))


has_text_type = URIRef(n["has_text_type"])
g.add((has_text_type, RDF.type, OWL.ObjectProperty))

has_genre = URIRef(n['has_genre'])

has_theme = URIRef(n['has_theme'])

has_subtheme = URIRef(n['has_subtheme'])


hasEra = URIRef(n['hasEra'])
g.add((hasEra, RDF.type, OWL.ObjectProperty))

Era = URIRef(n.Era)
Traditional = URIRef(n.Traditional)
Contemporary = URIRef(n.Contemporary)

g.add((Era, RDF.type, OWL.Class))
g.add((Traditional, RDF.type, OWL.Class))
g.add((Traditional, RDFS.subClassOf, Era))
g.add((Contemporary, RDF.type, OWL.Class))
g.add((Contemporary, RDFS.subClassOf, Era))

g.add((Text, hasEra, Era))
```

```python
# ontology creation
g.add((Text, RDF.type, OWL.Class))
g.add((Author, RDF.type, OWL.Class))
restriction = BNode()
g.add((restriction, RDF.type, OWL.Restriction))

g.add((Text_Complexity, RDF.type, OWL.Class))

g.add((Qualitative_Measures, RDF.type, OWL.Class))
g.add((Quantitative_Measures, RDF.type, OWL.Class))

g.add((Levels_of_Meaning, RDF.type, OWL.Class))
g.add((Text_Structure, RDF.type, OWL.Class))
g.add((Language_Conventionality_and_Clarity, RDF.type, OWL.Class))
g.add((Knowledge_Demands, RDF.type, OWL.Class))
g.add((Complexity_Measurement, RDF.type, OWL.Class))
 
g.add((Exceedingly_Complex, RDF.type, OWL.Class))
g.add((Very_Complex, RDF.type, OWL.Class))
g.add((Moderately_Complex, RDF.type, OWL.Class))
g.add((Slightly_Complex, RDF.type, OWL.Class))

g.add((Qualitative_Measures, RDFS.subClassOf, Text_Complexity))
g.add((Quantitative_Measures, RDFS.subClassOf, Text_Complexity))

g.add((Levels_of_Meaning, RDFS.subClassOf, Qualitative_Measures))
g.add((Text_Structure, RDFS.subClassOf, Qualitative_Measures))
g.add((Language_Conventionality_and_Clarity, RDFS.subClassOf, Qualitative_Measures))
g.add((Knowledge_Demands, RDFS.subClassOf, Qualitative_Measures))
g.add((Complexity_Measurement, RDFS.subClassOf, Qualitative_Measures))

g.add((Lexile_Level, RDFS.subClassOf, Quantitative_Measures))
g.add((Flesch_Kincaid, RDFS.subClassOf, Quantitative_Measures))
g.add((ATOS_Reading_Level, RDFS.subClassOf, Quantitative_Measures))
g.add((Reader_Maturity, RDFS.subClassOf, Quantitative_Measures))

g.add((Exceedingly_Complex, RDFS.subClassOf, Complexity_Measurement))
g.add((Very_Complex, RDFS.subClassOf, Complexity_Measurement))
g.add((Moderately_Complex, RDFS.subClassOf, Complexity_Measurement))
g.add((Slightly_Complex, RDFS.subClassOf, Complexity_Measurement))

g.add((has_text_complexity, RDF.type, OWL.ObjectProperty))
g.add((has_levels_of_meaning, RDF.type, OWL.ObjectProperty))
g.add((has_text_structure, RDF.type, OWL.ObjectProperty))
g.add((has_language_conventionality_and_clarity, RDF.type, OWL.ObjectProperty))
g.add((has_knowledge_demands, RDF.type, OWL.ObjectProperty))

has_author = URIRef(n['has_author'])
has_written = URIRef(n['has_written'])
g.add((has_author, RDF.type, OWL.ObjectProperty))
g.add((has_written, RDF.type, OWL.ObjectProperty))

g.add((has_theme, RDF.type, OWL.ObjectProperty))
g.add((has_genre, RDF.type, OWL.ObjectProperty))
g.add((has_subtheme, RDF.type, OWL.ObjectProperty))


g.add((has_author, RDFS.domain, Text))
g.add((has_author, RDFS.range, Author))

g.add((has_written, RDFS.domain, Author))
g.add((has_written, RDFS.range, Text))
g.add((has_author, OWL.inverseOf, has_written))

g.add((Text, has_author, Author))
g.add((Author, has_written, Text))


```

```python
import sys
import os

sys.path.append(os.path.abspath(".."))
from utils.utils import clean_text
```

```python
text_complexity_mapping = {
    'slightly complex': Slightly_Complex,
    'very complex': Very_Complex,
    'moderately complex': Moderately_Complex,
    'exceedingly complex': Exceedingly_Complex,
}
```

```python
# data population
counter = 0
genre_counter = 0
genre_failure_counter = 0
theme_counter = 0
theme_failure_counter = 0
subtheme_counter = 0
subtheme_failure_counter = 0

for index, row in df.iterrows():
    text_type_value = str(row['text type']).strip().lower().replace(' ', '_').replace('/','_').replace('-','_')
    if pd.notnull(row['text type']) and (text_type_value not in ['documentary', 'multimodal_tedtalk', 'poem_short_film', 'poem_song', 'short_film', 'video']):
        counter += 1
        text_name = clean_text(str(row['book']))
        text_uri = URIRef(n[text_name])
        g.add((text_uri, RDF.type, Text))

        if pd.notnull(row['author']):
            author_name = clean_text(str(row['author']))
            author_uri = URIRef(n[author_name])
            g.add((author_uri, RDF.type, Author))

            g.add((text_uri, URIRef(n['has_author']), author_uri))
            g.add((author_uri, URIRef(n['has_written']), text_uri))

        if pd.notnull(row['length (pages, time stamp, slide #) ']):
            g.add((text_uri, has_pages, Literal(row['length (pages, time stamp, slide #) '], datatype=XSD.numeric)))

        if pd.notnull(row['text type']):
            g.add((text_uri, has_text_type, text_type_mapping[str(row['text type']).strip().lower().replace(' ', '_').replace('/','_').replace('-','_')]))

        if pd.notnull(row['summary']):
            g.add((text_uri, has_summary, Literal(row['summary'],datatype=XSD.string)))
    
        # year of publication
        if pd.notnull(row['year of publication']):
            g.add((text_uri, has_year_of_publication, Literal(row['year of publication'], datatype=XSD.gYear)))
            if 'BC' in str(row['year of publication']):
                g.add((text_uri, hasEra, Traditional))
            elif (int(row['year of publication']) < 1945):
                g.add((text_uri, hasEra, Traditional))
            else:
                g.add((text_uri, hasEra, Contemporary))

        if pd.notnull(row['levels of meaning']):
            g.add((text_uri, has_levels_of_meaning, text_complexity_mapping[row['levels of meaning'].strip().lower()]))

        if pd.notnull(row['text structure']):
            g.add((text_uri, has_text_structure, text_complexity_mapping[row['text structure'].strip().lower()]))

        if pd.notnull(row['language conventionality and clarity']):
            g.add((text_uri, has_language_conventionality_and_clarity, text_complexity_mapping[row['language conventionality and clarity'].strip().lower()]))

        if pd.notnull(row['knowledge demands']):
            g.add((text_uri, has_knowledge_demands, text_complexity_mapping[row['knowledge demands'].strip().lower()]))                

        if pd.notnull(row['lexile level']) and (pd.notnull(row['levels of meaning']) or pd.notnull(row['text structure']) or pd.notnull(row['language conventionality and clarity']) or pd.notnull(row['knowledge demands'])):

            text_structure = row['text structure'].strip().lower()
            levels_of_meaning = row['levels of meaning'].strip().lower()
            language_conventionality_and_clarity = row['language conventionality and clarity'].strip().lower()
            knowledge_demands = row['knowledge demands'].strip().lower()

            lexile_level = str(row['lexile level'])
            if '-' in lexile_level:
                lexile_level_min, lexile_level_max = lexile_level.split('-')
                lexile_level_min = int(lexile_level_min[:-1])
                lexile_level_max = int(lexile_level_max[:-1])
                lexile_level_value = (lexile_level_min + lexile_level_max) / 2
            else:
                lexile_level_value = int(lexile_level)
            

            # text complexity
            if (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'slightly complex') and lexile_level_value<925:
                g.add((text_uri, has_text_complexity, Slightly_Complex))
            elif (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'moderately complex') and lexile_level_value>=925 and lexile_level_value<1185:
                g.add((text_uri, has_text_complexity, Moderately_Complex))
            elif (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'very complex') and lexile_level_value>=1185 and lexile_level_value<1335:
                g.add((text_uri, has_text_complexity, Very_Complex))
            elif (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'exceedingly complex') and lexile_level_value>=1335 and lexile_level_value<1440:
                g.add((text_uri, has_text_complexity, Exceedingly_Complex))


        # Quantitative measures
        if pd.notnull(row['lexile level']):
            g.add((text_uri, has_lexile_level, Literal(row['lexile level'], datatype=XSD.numeric)))
        
        if pd.notnull(row['Flesch-Kincaid']):
            g.add((text_uri, has_flesch_kincaid, Literal(row['Flesch-Kincaid'], datatype=XSD.decimal)))

        if pd.notnull(row['ATOS Reading Level']):
            g.add((text_uri, has_atos_reading_level, Literal(row['ATOS Reading Level'], datatype=XSD.decimal)))

        if pd.notnull(row['reader maturity']):
            reader_maturity = row['reader maturity'].strip()
            reader_maturity_uri = URIRef(n[clean_text(reader_maturity)])

            if reader_maturity == 'adult':
                lower_value = '13'
                upper_value = '99'
            else:
                lower_value, upper_value = reader_maturity.split(' to ')
                lower_value = 0 if lower_value.lower() == 'k' else int(lower_value)
                upper_value = int(upper_value)
            
            g.add((text_uri, has_reader_maturity, reader_maturity_uri))
            g.add((reader_maturity_uri, RDFS.label, Literal(reader_maturity, datatype=XSD.string)))
            g.add((reader_maturity_uri, has_min_reading_maturity, Literal(lower_value, datatype=XSD.numeric)))
            g.add((reader_maturity_uri, has_max_reading_maturity, Literal(upper_value, datatype=XSD.numeric)))



        # Genre
        if pd.notnull(row['genre']):
            genres = row['genre'].split(',')
            for val in genres:
                cleaned_val = val.strip().lower()
                if genre.get_genre_mapping(cleaned_val):
                    g.add((text_uri, has_genre, genre.get_genre_mapping(cleaned_val)))   
                    genre_counter += 1         
                else:
                    # print(cleaned_val)
                    genre_failure_counter += 1

    
        # Theme
        if pd.notnull(row['theme/topic']):
            themes = row['theme/topic'].split(',')
            for val in themes:
                cleaned_val = val.lower().replace("'",'').replace('-',' ').replace('_', ' ').replace('.','').strip()
                if theme.get_theme_mapping(cleaned_val):
                    g.add((text_uri, has_theme, theme.get_theme_mapping(cleaned_val)))
                    theme_counter += 1
                else:
                    # print(f'{text_name}: {cleaned_val}')
                    # print(f'{cleaned_val}')
                    theme_failure_counter += 1
    
        # Subthemes
        if pd.notnull(row['subgenre']):
            subgenres = row['subgenre'].split(',')
            for val in subgenres:
                cleaned_val = val.lower().replace("'",'').replace('-',' ').replace('_', ' ').replace('.','').strip()
                if subtheme.get_subtheme_mapping(cleaned_val):
                    g.add((text_uri, has_subtheme, subtheme.get_subtheme_mapping(cleaned_val)))
                    subtheme_counter += 1
                else:
                    # print(f'{text_name}: {cleaned_val}')
                    # print(f'{cleaned_val}')
                    subtheme_failure_counter += 1
```

```python
print(f'Total number of Texts processed: {counter}')
print(f'Total number of Genres mapped: {genre_counter}')
print(f'Total number of Genres failed to map: {genre_failure_counter}')
print(f'Total number of Themes mapped: {theme_counter}')
print(f'Total number of Themes failed to map: {theme_failure_counter}')
print(f'Total number of Subthemes mapped: {subtheme_counter}')
print(f'Total number of Subthemes failed to map: {subtheme_failure_counter}')
```
