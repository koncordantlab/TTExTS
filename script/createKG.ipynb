```python
def add_restriction (g, source_class_name, property_name, dest_class_name):
    restriction = BNode()
    g.add((restriction, RDF.type, OWL.Restriction))
    g.add((restriction, OWL.onProperty, property_name))
    g.add((restriction, OWL.someValuesFrom, dest_class_name))
    g.add((source_class_name, RDFS.subClassOf, restriction))
    return g
```

```python
def check_qualitative_measure(book_uri, graph, row, row_name):
    import pandas as pd
    if pd.notnull(row[row_name]):
        prop = predicates_ns['has_'+row_name.replace(' ','_')]
        match_case = row[row_name].strip().lower()
        if match_case == 'exceedingly complex':
            graph = add_restriction(graph, book_uri, prop, Exceedingly_Complex)
            # graph.add((book_uri, predicates_ns['has_'+row_name.replace(' ','_')], Exceedingly_Complex ))
        elif match_case == 'very complex':
            graph = add_restriction(graph, book_uri, prop, Very_Complex)
            # graph.add((book_uri, predicates_ns['has_'+row_name.replace(' ','_')], Very_Complex ))
        elif match_case == 'moderately complex':
            graph = add_restriction(graph, book_uri, prop, Moderately_Complex)
            # graph.add((book_uri, predicates_ns['has_'+row_name.replace(' ','_')], Moderately_Complex ))
        elif match_case == 'slightly complex':
            graph = add_restriction(graph, book_uri, prop, Slightly_Complex)
            # graph.add((book_uri, predicates_ns['has_'+row_name.replace(' ','_')], Slightly_Complex ))
    return graph
```

```python
# data population
for index, row in df.iterrows():
    book_name = str(row['book']).strip().lower().replace("'","").replace(" ","_").replace("-","")
    book_uri = URIRef(book_ns[book_name])
    author_name = str(row['author']).strip().lower().replace("'","").replace(" ","_").replace("-","")
    author_uri = URIRef(author_ns[author_name])
    
    # basic properties of book
    # book and author
    g.add((book_uri, RDF.type, OWL.Class))
    g.add((book_uri, RDFS.subClassOf, Book))
    g.add((author_uri, RDF.type, OWL.Class))
    g.add((author_uri, RDFS.subClassOf, Author))

    g = add_restriction(g, book_uri, has_author,author_uri)
    g = add_restriction(g, author_uri, has_written, book_uri)

    # texttype
    if pd.notnull(row['text type']):
        cleaned_val = row['text type'].strip().lower()
        g = add_restriction(g, book_uri, has_text_type, text_type_mapping[cleaned_val])

    # pages
    if pd.notnull(row['pages']):
        g.add((book_uri, has_pages, Literal(row['pages'], datatype=XSD.numeric)))
    
    # summary
    if pd.notnull(row['summary']):
        g.add((book_uri, has_summary, Literal(row['summary'],datatype=XSD.string)))
    
    # year of publication
    if pd.notnull(row['year of publication']):
        g.add((book_uri, has_year_of_publication, Literal(row['year of publication'], datatype=XSD.gYear)))
        if row['year of publication'] < 1945:
            g.add((book_uri, hasEra, Traditional))
        else:
            g.add((book_uri, hasEra, Contemporary))

    # Qualitative measures
    g = check_qualitative_measure(book_uri, g, row, 'text structure')
    g = check_qualitative_measure(book_uri, g, row, 'levels of meaning')
    g = check_qualitative_measure(book_uri, g, row, 'language conventionality and clarity')
    g = check_qualitative_measure(book_uri, g, row, 'knowledge demands')

    text_structure = row['text structure'].strip().lower()
    levels_of_meaning = row['levels of meaning'].strip().lower()
    language_conventionality_and_clarity = row['language conventionality and clarity'].strip().lower()
    knowledge_demands = row['knowledge demands'].strip().lower()
    
    # text complexity
    if (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'slightly complex') and row['lexile level']<925:
        g.add((book_uri, has_text_complexity, Slightly_Complex))
    elif (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'moderately complex') and row['lexile level']>=925 and row['lexile level']<1185:
        g.add((book_uri, has_text_complexity, Moderately_Complex))
    elif (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'very complex') and row['lexile level']>=1185 and row['lexile level']<1335:
        g.add((book_uri, has_text_complexity, Very_Complex))
    elif (text_structure or levels_of_meaning or language_conventionality_and_clarity or knowledge_demands == 'exceedingly complex') and row['lexile level']>=1335 and row['lexile level']<1440:
        g.add((book_uri, has_text_complexity, Exceedingly_Complex))


    # Quantitative measures
    if pd.notnull(row['lexile level']):
        g.add((book_uri, has_lexile_level, Literal(row['lexile level'], datatype=XSD.numeric)))
    
    if pd.notnull(row['Flesch-Kincaid']):
        g.add((book_uri, has_flesch_kincaid, Literal(row['Flesch-Kincaid'], datatype=XSD.decimal)))

    if pd.notnull(row['ATOS Reading Level']):
        g.add((book_uri, has_atos_reading_level, Literal(row['ATOS Reading Level'], datatype=XSD.decimal)))

    if pd.notnull(row['reader maturity']):
        reader_maturity = row['reader maturity'].strip().split()[0]
        lower_value = reader_maturity.split('-')[0]
        higher_value = reader_maturity.split('-')[1][:-2]
        g.add((book_uri, has_min_reading_maturity, Literal(lower_value, datatype=XSD.numeric)))
        g.add((book_uri, has_max_reading_maturity, Literal(higher_value, datatype=XSD.numeric)))
    


    # Genre
    if pd.notnull(row['genre']):
        genres = row['genre'].split(',')
        for val in genres:
            cleaned_val = val.strip().lower()
            try:
               g = add_restriction(g, book_uri, has_genre, genre_mapping[cleaned_val])            
            except KeyError:
                print(cleaned_val)

    
    # Theme
    if pd.notnull(row['theme/topic']):
        themes = row['theme/topic'].split(',')
        for val in themes:
            cleaned_val = val.lower().replace('-',' ').replace('_', ' ').replace('.','').strip()
            try:
                g = add_restriction(g, book_uri, has_theme, theme_mapping[cleaned_val])        
            except KeyError:
                print(val)
    
    # Subthemes
    if pd.notnull(row['subgenre']):
        subgenres = row['subgenre'].split(',')
        for val in subgenres:
            cleaned_val = val.lower().replace('-',' ').replace('_', ' ').strip()
            try:
                g = add_restriction(g, book_uri, has_subtheme, subgenre_mapping[cleaned_val])
            except KeyError:
                print(val)
```
